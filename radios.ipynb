```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):

    r1=1
    r2=0.001
    a=1
    b=1e-1

    if abs(x)<r2:
        return a*(1-np.exp(-b*(2*(np.sqrt(r1**2-x**2)-np.sqrt(r2**2-x**2)))))
    else:
        return a*(1-np.exp(-b*(2*(np.sqrt(r1**2-x**2)))))

def f2(x):

    r1=1
    r2=0.001
    a=1
    b=20

    if abs(x)<r2:
        return a*(1-np.exp(-b*(2*(np.sqrt(r1**2-x**2)-np.sqrt(r2**2-x**2)))))
    else:
        return a*(1-np.exp(-b*(2*(np.sqrt(r1**2-x**2)))))
def f3(x):

    r1=1
    r2=0.5
    a=1
    b=1e-1

    if abs(x)<r2:
        return a*(1-np.exp(-b*(2*(np.sqrt(r1**2-x**2)-np.sqrt(r2**2-x**2)))))
    else:
        return a*(1-np.exp(-b*(2*(np.sqrt(r1**2-x**2)))))

def f4(x):

    r1=1
    r2=0.5
    a=1
    b=20

    if abs(x)<r2:
        return a*(1-np.exp(-b*(2*(np.sqrt(r1**2-x**2)-np.sqrt(r2**2-x**2)))))
    else:
        return a*(1-np.exp(-b*(2*(np.sqrt(r1**2-x**2)))))





x= np.linspace(-1.,1.,100)
y = [f(i) for i in x]
y2= [f2(i) for i in x]
y3 = [f3(i) for i in x]
y4= [f4(i) for i in x]

plt.plot(x, y, label= r'b=0.1, $R_2$=0.001')
plt.plot(x, y2, label = r'b=2, $R_2$=0.001')
plt.plot(x, y3, label = r'b=0.1, $R_2$=0.5')
plt.plot(x, y4, label = r'b=2, $R_2$=0.5')
plt.xlabel("x")
plt.ylabel(r"$\mathcal{I}$(x)")
plt.title(r"Gráfica de $\mathcal{I}$(x)")
plt.legend(loc='upper right', bbox_to_anchor=(.9, .5, 0.5, 0.5))
plt.grid(True)
plt.show()
```
