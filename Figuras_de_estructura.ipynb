```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse, Polygon, FancyArrow, Arc
import matplotlib.image as mpimg

```

```python
fig, ax = plt.subplots(figsize=(5,4))

# Configuración del eje
ax.set_xlim(-2, 2)
ax.set_ylim(0, 5.4)
ax.set_aspect('equal')
ax.axis("off")

# Funciones para las curvas laterales
z = np.linspace(0, 5, 200)
radius_top = 1.5
radius_bottom = 0.45
radius = radius_bottom + (radius_top - radius_bottom)*(z/5)**0.3

ax.plot(radius, z, 'k', linewidth=2)
ax.plot(-radius, z, 'k', linewidth=2)

# Funciones para las curvas laterales (internas)
radius_top_in = .9
radius_bottom_in = 0.2
radius_in = radius_bottom_in + (radius_top_in - radius_bottom_in)*(z/5)**0.3


ax.plot(radius_in, z, 'k', linewidth=2)
ax.plot(-radius_in, z, 'k', linewidth=2)

# Funciones para sombreado lateral
radius_top_colored = np.linspace(.9, 1.5, 200)
radius_bottom_colored = np.linspace(0.2, 0.45, 200)
radius_=np.zeros(200)
for i in range(0,200):
    radius_ = radius_bottom_colored[i] + (radius_top_colored[i] - radius_bottom_colored[i])*(z/5)**0.3

    ax.plot(radius_, z, linewidth=2, color='blue', alpha=0.1)
    ax.plot(-radius_, z, linewidth=2, alpha=0.1, color='blue')


# Discos horizontales (z0, zk, z_inj,max, z_max)
levels = [.75, 2.5, 3.5, 5]
for lv in levels:
    r_lv = radius_bottom + (radius_top - radius_bottom)*(lv/5)**0.3
    
    ell = Ellipse((0, lv), 2*r_lv, 0.5*r_lv, fill=False, linestyle='--', edgecolor='black' ,zorder=5)
    
    ax.add_patch(ell)

# Disco horizontal interno
r_lv_in = radius_bottom_in + (radius_top_in - radius_bottom_in)*(0.75/5)**0.3
ell_in = Ellipse((0, 0.75), 2*r_lv_in, 0.5*r_lv_in, fill=False, linestyle='--', edgecolor='black' ,zorder=5)
ax.add_patch(ell_in)

# Disco horizontal interno z_m_iny
# r_lv_in = radius_bottom_in + (radius_top_in - radius_bottom_in)*(3.5/5)**0.3
# ell_in = Ellipse((0, 3.5), 2*r_lv_in, 0.5*r_lv_in, fill=False, linestyle='--', edgecolor='black' ,zorder=5)
# ax.add_patch(ell_in)

# Región sombreada alrededor de zk
zk = 2.5
r_zk = radius_bottom + (radius_top - radius_bottom)*(zk/5)**0.3
shade = Ellipse((0, zk), 2*r_zk, 0.5*r_zk, fill=True, alpha=0.3, color='gray')
ax.add_patch(shade)

# Región sombreada borde superior
zk=5
r_zk = radius_bottom + (radius_top - radius_bottom)*(zk/5)**0.3
shade_in = Ellipse((0, zk), 2*r_zk, 0.5*r_zk, fill=True, alpha=1, color='white', zorder=4)
ax.add_patch(shade_in)


# # Flechas ascendentes
# for x in [-0.4, 0, 0.4]:
#     ax.add_patch(FancyArrow(x, 1.5, 0, 3.5, width=0.03, color='blue'))

# # Ángulo θ
# ax.plot([0, 1.5], [2.5, 4.5], color='green', linewidth=2)
# ax.text(0.4, 3.4, r'$\theta$', color='green', fontsize=16)

# Etiquetas
ax.text(1.5, 4.5, r'$z_{\max}$', fontsize=14)
ax.text(1.5, 3.5, r'$z_{\mathrm{inj,max}}$', fontsize=14)
# ax.text(1.4, 2.5, r'$z_k$', fontsize=14)
ax.text(1.4, .75, r'$z_0$', fontsize=14)
ax.text(-2, 2.5, r'$R_{in}$', fontsize=14, color='C0',zorder=6)
ax.text(1.5, 2.5, r'$R_{out}$', fontsize=14, color='red',zorder=6)

# # Agujero negro en la base
# ax.add_patch(plt.Circle((0,0), 0.25, color='black'))
# ax.text(0.4, 0, "SMBH", fontsize=14)

#Hlines R_in R_out
lv=2.5
r_lv = radius_bottom + (radius_top - radius_bottom)*(lv/5)**0.3
r_lv_in = radius_bottom_in + (radius_top_in - radius_bottom_in)*(lv/5)**0.3
ax.hlines(lv,0.0,r_lv,ls='--', color='red')
ax.hlines(lv,-r_lv_in,0.0,ls='--')

# Eje z
ax.arrow(0, 0, 0, 4.7, width=0.01, head_width=0.15, color='black',zorder=10)
ax.text(0.1, 5, r'$z$', fontsize=16, zorder=10)
plt.savefig('Limbed_jet.png', dpi=300, transparent=True)
plt.show()
```
