```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import smplotlib
```

```python
z_proj, r = np.loadtxt('../ObsData/jet_diameter.dat', skiprows=2, unpack=True)
z_proj *= 2
r *= 2
z = z_proj / np.sin(np.deg2rad(17))

def r_fun1(z):
    g1 = 0.3
    g2 = 0.58
    z0 = 4.
    r0 = 10.
    return r0 * np.power(z/z0, g2)

def r_fun2(z):
    g = 0.15
    z0 = 4.
    r0 = 40
    return r0 * np.power(z/z0, g)

def r_fun(z):
    return np.fmax(r_fun1(z),r_fun2(z))
```

```python
z_vec = np.geomspace(2, 1e4, 1000)

fig, ax = plt.subplots()

ax.set_xlabel(r'$z ~ [R_{\rm g}]$')
ax.set_ylabel(r'$r ~ [R_{\rm g}]$')
ax.loglog(z, r, ls='', marker='s')
ax.loglog(z_vec, r_fun(z_vec))
```
